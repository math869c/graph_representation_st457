```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("training_curves.csv")
```

```python
plt.figure(figsize=(8,5))

df_gat = df[df["Model"] == "GAT"]

for edge, group in df_gat.groupby("Edge Type"):
    plt.plot(group["Epoch"], group["Val Loss"], label=f"GAT-{edge}")

plt.xlabel("Epoch")
plt.ylabel("Validation Loss")
plt.title("GAT Validation Loss by Edge Type")
plt.legend()
plt.grid(alpha=0.3)
plt.tight_layout()

plt.savefig("val_loss_gat.png", dpi=300)
plt.show()
```

```python
plt.figure(figsize=(8,5))

df_rotate = df[df["Model"] == "GAT+RotatE"]

for edge, group in df_rotate.groupby("Edge Type"):
    plt.plot(group["Epoch"], group["Val Loss"], label=f"GAT+RotatE-{edge}")

plt.xlabel("Epoch")
plt.ylabel("Validation Loss")
plt.title("GAT+RotatE Validation Loss by Edge Type")
plt.legend()
plt.grid(alpha=0.3)
plt.tight_layout()

plt.savefig("val_loss_gat_rotate.png", dpi=300)
plt.show()
```
